
## FISH ANALYSIS USING LENGTH1

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Step 1: Load the dataset

```python

df = pd.read_csv("Fish.csv")

# Step 2: Understand the dataset
print("First five rows:")
print(df.head())

print("\nDataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

## Step 3: Check data quality

```python

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

## Step 4: Identify fish classes

```python
# Step 4: Identify fish classes
print("\nFish species/classes:")
print(df["Species"].unique())

print("\nNumber of classes:", df["Species"].nunique())

print("\nRecords per class:")
print(df["Species"].value_counts())
```

```python
# Step 5: Check the selected feature
print("\nLength1 statistics:")
print(df["Length1"].describe())
```

## One-dimensional visualization

```python
# Step 6: One-dimensional visualization
plt.figure(figsize=(12, 6))

species_list = sorted(df["Species"].unique())

for i, species in enumerate(species_list):
    values = df.loc[df["Species"] == species, "Length1"]
    plt.scatter(values, [i] * len(values), alpha=0.7,
                label=species)

plt.yticks(range(len(species_list)), species_list)
plt.xlabel("Length1")
plt.ylabel("Fish Species")
plt.title("One-Dimensional Analysis Using Length1")
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Step 7: Compare length statistics by class
print("\nLength1 statistics by species:")
print(
    df.groupby("Species")["Length1"]
      .agg(["count", "mean", "min", "max"])
      .round(2)
)
```
